## 8. What Are Customers Actually Complaining About?

### Question 8
What are the most common complaint keywords in critical reviews?

```python
import re
```

```python
re.sub(r'[^a-zA-Z\s]', '', "Hello! 123")
```

```python
# re is used to clean text with patterns, like removing punctuation and numbers.
import re
from collections import Counter

# Cleans review text
critical['Clean Review'] = (
    critical['Review Text']
    .fillna('')
    .str.lower()
    .apply(lambda x: re.sub(r'[^a-zA-Z\s]', '', x))
)

# to count reviews
words = '  '.join(critical['Clean Review']).split()

stop_words = {
    'the','and','is','was','to','a','of','in','for',
    'this','it','i','my','on','with','that','but',
    'very','have','had','not'
}

words = [word for word in words if word not in stop_words]

word_counts = Counter(words)

word_counts.most_common(15)

# For Visualizatoin
top_words = pd.DataFrame(
    word_counts.most_common(15),
    columns=['Word','Frequency']
)

top_words.sort_values('Frequency').plot(
    x='Word',
    y='Frequency',
    kind='barh',
    figsize=(9,6),
    legend=False,
    title='Top Complaint Keywords in Critical Reviews'
)

plt.xlabel('Frequency')
plt.ylabel('Keyword')
plt.show()
```

### Insight:
The most frequent complaint terms reveal the specific areas customers mention when describing negative experiences. These keywords can be grouped into broader issue categories for actionable analysis.

## 9.Advanced: What Type of Complaint Is Most Common?

```python
# Instead of analyzing individual words, create business categories:
complaints = {
    'Fit & Size': [
        'size', 'fit', 'small', 'large', 'tight', 'loose'
    ],
    'Quality': [
        'quality', 'material', 'fabric', 'cheap', 'thin'
    ],
    'Color': [
        'color', 'colour', 'shade'
    ],
    'Delivery': [
        'shipping', 'delivery', 'arrived', 'late'
    ],
    'Return & Refund': [
        'return', 'refund', 'exchange'
    ]
}

# Count them
complaint_counts = {}

for category, keywords in complaints.items():
    complaint_counts[category] = critical['Clean Review'].apply(
        lambda text: any(
            keyword in text.split()
            for keyword in keywords
        )
    ).sum()

complaint_counts

# For Visualization
pd.Series(complaint_counts).sort_values().plot(
    kind='barh',
    figsize=(9,5),
    title='Major Complaint Categories'
)

plt.xlabel('Number of Critical Reviews')
plt.ylabel('Complaint Category')
plt.show()
```

### Insight:
Customers' negative experiences can be grouped into major categories such as fit & size, product quality, color, delivery, and returns. This makes the raw review data easier for the business to act upon.

## 10.Do Unhappy Customers Write Longer Reviews?
- Based on EDA

### Question 10
Is review length associated with customer rating?

```python
Women_clothing['Review Length'] = Women_clothing['Review Text'].fillna('').str.len()
avg_length = Women_clothing.groupby('Rating')['Review Length'].mean()
avg_length

# For Visualization
avg_length.plot(
    kind='bar',
    figsize=(8,5),
    title='Average Review Length by Rating'
)

plt.xlabel('Rating')
plt.ylabel('Average Review Length')
plt.xticks(rotation=0)
plt.show()
```

## 11.Does Recommendation Match Rating?

### Question 11
Are customers who give low ratings also less likely to recommend the product?

```python
# Recommendation percent by rating
recommendation = pd.crosstab(
    Women_clothing['Rating'],
    Women_clothing['Recommended IND'],
    normalize='index'
) * 100

recommendation

# For Visualization
recommendation.plot(
    kind='bar',
    figsize=(9,5),
    title='Recommendation Rate by Customer Rating'
)

plt.xlabel('Rating')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='Recommended')
plt.show()
```

### Insight:
This analysis examines whether star ratings are consistent with customers' willingness to recommend the product.
